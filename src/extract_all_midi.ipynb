```python
import os
from extract_full_midi import extract_full_song

#++++++++++++++++++++++++++++++++++++++++++++++++++++++++
#+   Extract all the MIDI files from the base_folders   +
#++++++++++++++++++++++++++++++++++++++++++++++++++++++++

# List of base sample folders
base_folders = [
    "/workspace/samples/lastfm_samples",
    "/workspace/samples/suno_samples",
    "/workspace/samples/udio_samples"
]

# Loop over each sample folder
for base in base_folders:
    segmented_dir = os.path.join(base, "segmented")
    if not os.path.isdir(segmented_dir):
        print(f"Segmented folder not found in {base}")
        continue

    # For each song folder in the segmented folder
    for song_folder in os.listdir(segmented_dir):
        song_path = os.path.join(segmented_dir, song_folder)
        if not os.path.isdir(song_path):
            continue

        # Locate the audio folder inside the song folder
        audio_folder = os.path.join(song_path, "audio")
        if not os.path.isdir(audio_folder):
            print(f"Audio folder not found in {song_path}")
            continue

        # Create a midi folder inside the song folder if it doesn't exist
        midi_folder = os.path.join(song_path, "midi")
        os.makedirs(midi_folder, exist_ok=True)

        # Iterate over all .wav files in the audio folder
        for file in os.listdir(audio_folder):
            if file.lower().endswith(".wav"):
                audio_path = os.path.join(audio_folder, file)
                base_name = os.path.splitext(file)[0]
                output_midi_path = os.path.join(midi_folder, f"{base_name}.mid")
            
                try:
                    extract_full_song(audio_path, output_midi_path)
                except Exception as e:
                    print(f"Error processing {audio_path}: {e}")

print("Extraction complete!")
```
