```python
#Check Plagiarism
file = open('../data/chord_progressoins_new.txt', 'r')
lines = file.readlines()
sample = []
for line in lines:
    e = line.split(',')
    e[-1] = e[-1].replace('\n', '')
    sample.append(e)

print(len(sample))
```

```python
trigramSearch=[]
for x in range(len(sample)):
    ts = []
    for y in range(len(sample[x])-2):
        tri = (sample[x][y], sample[x][y+1], sample[x][y+2])
        ts.append(tri)
    trigramSearch.append(ts)
```

```python
#start an array of strings of chords
def stringSongs(song):
    a = song[np.where(song != '<pad>')]
    r = " ".join(a)
    return r

strSong = []
for song in all_chords_extended:
    r = stringSongs(song)
    strSong.append(r)
```

```python
print(strSong[0])
```

```python
def checkChordCorpus(search, strSong):
    i = 0
    counter = 0
    result = []
    for chord in strSong:
        #print(song, end=' ')
        if search in chord:
            r = {}
            r['composer'] = extended_meta[i]['composer']
            r['song'] = extended_meta[i]['song_name']
            result.append(r)
            counter += 1
        i += 1
    return result, counter
```

```python
search = 'Dm7'

r,c = checkChordCorpus(search, strSong)
print(c)
```

```python
result = []
for i in tqdm(range(len(trigramSearch))):
    line = 0
    for tri in trigramSearch[i]:
        s = ''.join(tri)
        found, counter = checkChordCorpus(s, strSong)
        if counter > 0:
            a = {}
            a['sequence'] = i
            a['id'] = line
            a['trigram'] = tri
            a['found'] = found
            a['count'] = counter
            result.append(a)
        line += 1
result = np.array(result, dtype=object)
```
